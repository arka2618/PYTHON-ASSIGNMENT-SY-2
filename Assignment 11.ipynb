```python
import numpy as np
import pandas as pd
```

## Create Series

```python
num = np.random.randint(low=5, high=20, size=10)
num
```

```python
df1 = pd.DataFrame({'Num' : num})
df1
```

```python
df1.shape
```

```python
df1.index
```

```python
df1.columns
```

## Perform Indexing

```python
#First row
df1.iloc[1]
```

```python
#Slicing row
df1.iloc[:5]
```

```python
#Get specific value
df1.loc[4, 'Num']
```

## Filtering

```python
df1[df1['Num'] > 15]
```

```python
df1[df1['Num'] % 2 == 0]
```

## Statistical Operations

```python
print(f"Mean: {df1['Num'].mean()}")
print(f"Maximum: {df1['Num'].max()}")
print(f"Minimum: {df1['Num'].min()}")
print(f"Median: {df1['Num'].median()}")
```

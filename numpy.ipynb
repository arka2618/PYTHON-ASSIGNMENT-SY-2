```python
import numpy as np
```

```python
temp = np.array([22.5, 24.0, 23.8, 25.1, 26.4, 24.8, 23.1, 22.9, 25.0, 26.2])
temp
```

```python
temp2 = temp[:3]
print(f"Temperature of first 3 days")
for i, j in zip(range(1, 4), temp2):
    print(f"Day{i}: {j}")
```

```python
temp3 = temp[4:8]
print(f"Temperature from day 5 to day 8")
for i, j in zip(range(5, 9), temp3):
    print(f"Day{i}: {j}")
```

```python
print(f"Average Temperature: {temp.mean()}")
print(f"Maximum Temperature: {temp.max()}")
print(f"Minimum Temperature: {temp.min()}")
print(f"Total Temperature: {temp.sum()}")
```
